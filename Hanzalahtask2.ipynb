```python
"""Question 1: Positive, Negative or Zero"""
num = float(input("Enter a number: "))
if num > 0:
    print(f"{num} is positive")
elif num < 0:
    print(f"{num} is negative")
else:
    print(f"{num} is zero")
```

```python
"""Question 2: Larger of Two Numbers"""
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
if a > b:
    print(f"{a} is larger")
else:
    print(f"{b} is larger")
```

```python
"""Question 3: Largest of Three Numbers"""
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
if a >= b and a >= c:
    print(f"{a} is the largest")
elif b >= a and b >= c:
    print(f"{b} is the largest")
else:
    print(f"{c} is the largest")
```

```python
"""Question 5: Minor, Adult or Senior Citizen"""
age = int(input("Enter your age: "))
if age < 18:
    print(f"Age {age}: Minor")
elif age > 18 and age < 60:
    print(f"Age {age}: Adult")
else:
    print(f"Age {age}: Senior Citizen")
```

```python
"""Question 6: Even or Odd"""
num = int(input("Enter an integer: "))
if num % 2 == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

```python
"""Question 7: Calculator using Operator"""
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter an operator (+, -, x, /): ")
if operator == "+":
    print(f"{num1} + {num2} = {num1 + num2}")
elif operator == "-":
    print(f"{num1} - {num2} = {num1 - num2}")
elif operator == "x":
    print(f"{num1} x {num2} = {num1 * num2}")
elif operator == "/":
    if num2 != 0:
        print(f"{num1} / {num2} = {num1 / num2}")
    else:
        print("Division by zero is not allowed")
else:
    print("Invalid operator")
```

```python
"""Question 8: Range Check 20 (inclusive) - 40 (exclusive)"""
num = float(input("Enter a number: "))
if num >= 20 and num < 40:
    print(f"{num} lies in the range 20-40")
else:
    print(f"{num} does not lie in the range 20-40")
```

```python
"""Question 9: Divisibility by 2, 3 or Both"""
num = int(input("Enter an integer: "))
if num % 2 == 0 and num % 3 == 0:
    print(f"{num} is divisible by both 2 and 3")
elif num % 2 == 0:
    print(f"{num} is divisible by 2")
elif num % 3 == 0:
    print(f"{num} is divisible by 3")
else:
    print(f"{num} is not divisible by 2 or 3")
```

```python
"""Question 10: Pass or Fail"""
score = float(input("Enter your score: "))
if score > 60:
    print(f"Score {score}: Pass")
else:
    print(f"Score {score}: Fail")
```

```python
"""Question 11: Freezing, Moderate or Hot Temperature"""
temp = float(input("Enter the temperature in Celsius: "))
if temp < 0:
    print(f"{temp} degrees Celsius: Freezing")
elif temp > 0 and temp < 26:
    print(f"{temp} degrees Celsius: Moderate")
else:
    print(f"{temp} degrees Celsius: Hot")
```

```python
"""Question 12: Salary Tax Deduction (Single Slab)"""
salary = float(input("Enter your salary: "))
if salary > 600000:
    tax = salary * 0.05
    gross_salary = salary - tax
    print(f"Tax deducted: {tax}")
    print(f"Gross Salary: {gross_salary}")
else:
    print(f"Gross Salary: {salary}")
```

```python
"""Question 13: Salary Tax Deduction (Multiple Slabs)"""
salary = float(input("Enter your salary: "))
if salary > 2500000:
    tax = salary * 0.15
elif salary >= 1500000 and salary <= 2500000:
    tax = salary * 0.10
elif salary >= 500000 and salary < 1500000:
    tax = salary * 0.05
else:
    tax = 0

gross_salary = salary - tax
print(f"Tax deducted: {tax}")
print(f"Gross Salary: {gross_salary}")
```
