### Question 14 - For Loop

```python
discard_menu = ["fish", "qeema", "nihari"]
menu = []
menu_price = {}
menu_price_per_head = {}
total_guests = 300

for i in range(1, 6):
    dish = input(f"Enter dish {i}: ")

    if dish.lower() in discard_menu:
        print(f"{dish} is not allowed. Skipping.")
        continue

    menu.append(dish)
    price = float(input(f"Enter total price of {dish} for {total_guests} guests: "))
    menu_price[dish] = price
    menu_price_per_head[dish] = price / total_guests

total_menu_price = 0
for dish in menu_price:
    total_menu_price += menu_price[dish]

total_per_head_cost = 0
for dish in menu_price_per_head:
    total_per_head_cost += menu_price_per_head[dish]

print(f"Final menu: {menu}")
print(f"Menu prices: {menu_price}")
print(f"Per head cost of each dish: {menu_price_per_head}")
print(f"Total menu price: {total_menu_price}")
print(f"Total per head cost of complete menu: {total_per_head_cost}")
```

### Question 14 - While Loop

```python
discard_menu = ["fish", "qeema", "nihari"]
menu = []
menu_price = {}
menu_price_per_head = {}
total_guests = 300
i = 1

while i <= 5:
    dish = input(f"Enter dish {i}: ")
    i += 1

    if dish.lower() in discard_menu:
        print(f"{dish} is not allowed. Skipping.")
        continue

    menu.append(dish)
    price = float(input(f"Enter total price of {dish} for {total_guests} guests: "))
    menu_price[dish] = price
    menu_price_per_head[dish] = price / total_guests

total_menu_price = 0
j = 0
dish_keys = list(menu_price.keys())
while j < len(dish_keys):
    total_menu_price += menu_price[dish_keys[j]]
    j += 1

total_per_head_cost = 0
k = 0
head_keys = list(menu_price_per_head.keys())
while k < len(head_keys):
    total_per_head_cost += menu_price_per_head[head_keys[k]]
    k += 1

print(f"Final menu: {menu}")
print(f"Menu prices: {menu_price}")
print(f"Per head cost of each dish: {menu_price_per_head}")
print(f"Total menu price: {total_menu_price}")
print(f"Total per head cost of complete menu: {total_per_head_cost}")
```
